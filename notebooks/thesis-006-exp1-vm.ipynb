## Check experiment data progress

````python
import os
from pathlib import Path
exp_dir = Path('../data/raw/exp1-vm-degradation')
levels = [1, 2, 3, 5, 7, 10]
reps   = [1, 2, 3]
completed = []
for level in levels:
    for rep in reps:
        run_dir = exp_dir / f'L{level}' / f'run{rep}'
        done = run_dir.exists() and (run_dir / 'dagster_runs.csv').exists()
        status = 'DONE' if done else 'TODO'
        completed.append((level, rep, status))
done_count = sum(1 for _, _, s in completed if s == 'DONE')
print(f'Progress: {done_count}/18 batches complete\n')
for level, rep, status in completed:
    print(f'  L{level} rep{rep}: {status}')
````

## Validate data quality after completion

````python
import os, pandas as pd
from pathlib import Path
exp_dir = Path('../data/raw/exp1-vm-degradation')
if not any(exp_dir.rglob('dagster_runs.csv')):
    print('No data yet — run Experiment 1 first.')
else:
    for csv_path in sorted(exp_dir.rglob('dagster_runs.csv')):
        df = pd.read_csv(csv_path)
        runs = len(df)
        ok = (df['status'] == 'SUCCESS').sum() if 'status' in df.columns else '?'
        print(f'  {csv_path.relative_to(exp_dir)}: {runs} runs, {ok} succeeded')
````
