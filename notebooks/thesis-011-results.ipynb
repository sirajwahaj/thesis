## Generate VM results LaTeX table snippet

```python
import pandas as pd
from pathlib import Path
vm_path = Path('../data/processed/exp1_vm_summary.csv')
if not vm_path.exists():
    print('exp1_vm_summary.csv not found — run analysis first (THESIS-010)')
else:
    df = pd.read_csv(vm_path)
    print('% Auto-generated from data/processed/exp1_vm_summary.csv')
    print('% Paste into docs/chapters/04-results/exp1-vm-degradation.tex')
    print()
    print(r'\begin{table}[htbp]')
    print(r'  \centering')
    print(r'  \caption{VM execution metrics by concurrency level (n=3 per level)}')
    print(r'  \begin{tabular}{lrrrr}')
    print(r'    \toprule')
    print(r'    Level & Concurrent Jobs & Success Rate (\%) & Mean Time (s) & Std Dev (s) \\')
    print(r'    \midrule')
    level_map = {1: 'L1', 2: 'L2', 3: 'L3', 5: 'L4', 7: 'L5', 10: 'L6'}
    for _, row in df.iterrows():
        level_label = level_map.get(int(row.get('level', 0)), '?')
        jobs   = int(row.get('concurrent_jobs', row.get('level', 0)))
        sr     = f"{row.get('success_rate', 0)*100:.1f}" if row.get('success_rate', 1) <= 1 else f"{row.get('success_rate', 0):.1f}"
        mean_t = f"{row.get('mean_execution_time', 0):.1f}"
        std_t  = f"{row.get('std_execution_time', 0):.1f}"
        print(f'    {level_label} & {jobs} & {sr} & {mean_t} & {std_t} \\\\')
    print(r'    \bottomrule')
    print(r'  \end{tabular}')
    print(r'  \label{tab:exp1-results}')
    print(r'\end{table}')
```
