## SQ1 answer template

```python
import pandas as pd
from pathlib import Path
vm_path = Path('../data/processed/exp1_vm_summary.csv')
if not vm_path.exists():
    print('Run THESIS-010 first.')
else:
    df = pd.read_csv(vm_path)
    col = 'success_rate'
    if col in df.columns:
        sr = df[col]
        threshold = 0.95 if sr.max() <= 1 else 95
        below = df[df[col] < threshold]
        if len(below) > 0:
            row = below.iloc[0]
            level_map = {1: 'L1', 2: 'L2', 3: 'L3', 5: 'L4', 7: 'L5', 10: 'L6'}
            lv = level_map.get(int(row['level']), str(row['level']))
            sr_pct = row[col]*100 if threshold == 0.95 else row[col]
            mean_t = row.get('mean_execution_time', '??')
            std_t  = row.get('std_execution_time', '??')
            print('SQ1 ANSWER TEMPLATE:')
            print(f'SQ1 is answered at concurrency level {lv} ({int(row["level"])} concurrent jobs),')
            print(f'where the VM job success rate first dropped below the 95% reliability threshold,')
            print(f'recording {sr_pct:.1f}% success rate')
            print(f'(mean execution time: {mean_t:.1f}s +/- {std_t:.1f}s).')
        else:
            print('VM success rate never dropped below 95% -- experiment may need more levels.')
    else:
        print(f'Column "{col}" not found. Available:', list(df.columns))
```
