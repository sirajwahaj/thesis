# THESIS-005 — Master Experiment Runner
**Story Points:** 2 | **Status:** DONE

`run_experiment.sh` orchestrates all levels, metrics collection, export, and metadata.


## Acceptance Criteria
- [x] Takes `experiment` and `environment` arguments
- [x] Iterates L1-L6 (1, 2, 3, 5, 7, 10) with 3 reps each
- [x] Starts/stops metrics collectors per batch
- [x] Writes `metadata.json` per run
- [x] 60-second cooldown between batches
- [x] `--dry-run` mode
- [x] `--levels` flag for specific levels
- [x] Dry-run at L1 and L2 passes

## Dry-run validation

```python
import subprocess
for env, exp in [('k8s', 'exp2a'), ('vm', 'exp1')]:
    r = subprocess.run(
        ['bash', '../scripts/run_experiment.sh', exp, env, '--dry-run', '--levels', '1 2'],
        capture_output=True, text=True, timeout=30
    )
    icon = 'OK' if r.returncode == 0 else 'FAIL'
    print(f'[{icon}] dry-run {exp} {env}')
    print(r.stdout[:1000])
    if r.returncode != 0:
        print('STDERR:', r.stderr[:500])
    print()
```
