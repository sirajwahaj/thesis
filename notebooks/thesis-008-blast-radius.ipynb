## Parse blast radius results

```python
import pandas as pd
from pathlib import Path
for env in ['vm', 'k8s']:
    blast_dir = Path(f'../data/raw/exp2-kubernetes-isolation/part-b-blast-radius/{env}')
    csvs = list(blast_dir.rglob('blast_radius.csv'))
    if not csvs:
        print(f'{env}: No blast_radius.csv found yet')
        continue
    dfs = [pd.read_csv(c) for c in csvs]
    combined = pd.concat(dfs)
    print(f'\n{env} blast radius results:')
    print(combined.describe())
```
